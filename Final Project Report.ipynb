## Task Description
#### Python - Multi-Point Trajectory Tacking
The team decided to tackle the "multi-point trajectory tacking." The goal of this project was to have a 2R planar robot follow a sequence of 3 or more waypoints in a smooth motion, mimicking a real world task. The team implemented this by tracing a shape. More specifically, the program asks the user via a GUI whether it wants to trace a square, circle, star, or pick points. When the user clicks "Run Simulation" the robot traces the specified shape or prompts the user to pick points.

## Lessons Learned
The biggest lesson the team learned from the final project is just how useful GitHub is. We previously used GitHub in PA2, but very incorrectly. This time around, the team sat down before we started coding and dedicated some time to learn GitHub. This proved very beneficial because we were much more efficient in sharing code and keeping records of our previous versions. This allowed us to dive much deeper into the project and code more because we weren't just undoing code when something went wrong. Also, when we weren't meeting, it was extremely nice to just send a message that mentioned you updated something rather than sending a script back and forth. The team all agreed that we will definitely be using this in the future for all coding projects.
