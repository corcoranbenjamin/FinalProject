## Task Description
#### Python - Multi-Point Trajectory Tacking
The team decided to tackle the "multi-point trajectory tacking." The goal of this project was to have a 2R planar robot follow a sequence of 3 or more waypoints in a smooth motion, mimicking a real world task. The team implemented this by tracing a shape. More specifically, the program asks the user via a GUI whether it wants to trace a square, circle, star, or pick points. When the user clicks "Run Simulation" the robot traces the specified shape or prompts the user to pick points.

## Control Strategy

τ = Kp·(θ_d - θ_a) + Kd·(θ̇_d - θ̇_a) + G(θ)

The PD + Gravity controller tuning for each joint was achieved through trial and error. Without the gravity compensation term in the controller, it impossible to get the steady state error near zero, it kept drifting down due to gravity regardless of the selected values for Kp and Kd, so gravity compensation was added to avoid the need for an integral term. Once Kp was identified to limit overshoot, Kd was increased to reduce the oscillations until the steady state error was less than 1º.

### Dynamics and Control Functions

#### manipulatorEOM()
Implements the equations of motion for the 2R manipulator. At each time step the function extracts the current joint positions and velocities, which are used to compute the mass matrix, Coriolis/centrifugal terms, and gravity vector using the example shown in class and the example from HW4. The PD control torques are computed based on position and velocity error. Then dynamic equation is solved for joint accelerations, then the time derivatives of the state vector are returned.

#### odeint()
The dynamics are simulated using odeint(). The initial joint states, time span, and manipulatorEOM() are passed in and the actual joint positions and velocities are computed at each time step. The results are used for driving the animation and tracking error.

## Lessons Learned
The biggest lesson the team learned from the final project is just how useful GitHub is. We previously used GitHub in PA2, but very incorrectly. This time around, the team sat down before we started coding and dedicated some time to learn GitHub. This proved very beneficial because we were much more efficient in sharing code and keeping records of our previous versions.
